```python
x1 = 1
x2 = 0

w1 = 0.8
w2 = -0.4

bias = -0.2
```

```python
def step_activation(value):
    if value >= 0:
        return 1
    else:
        return 0
```

```python
z = (x1 * w1) + (x2 * w2) + bias
print("Weighted sum:", z)

output = step_activation(z)
print("Perceptron output:", output)
```

```python
def perceptron(x1, x2, w1, w2, bias):
    weighted_sum = (x1 * w1) + (x2 * w2) + bias
    step_output = step_activation(weighted_sum)
    return weighted_sum, step_output
```

```python
weighted_sum, prediction = perceptron(1, 0, 0.8, -0.4, -0.2)
print("Weighted sum:", weighted_sum)
print("Perceptron output:", prediction)
```
